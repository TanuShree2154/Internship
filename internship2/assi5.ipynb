```python
# 1) Create a CSV file for address book, CSV file should have column for name, address, mobile,
#  email. Insert 2-3 dummy data entered by user.

import csv
d=[
    'name','address','mobile','email'
]
with open('address_book','w',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(d)
    n=int(input("enter no. of inputs:"))
    for i in range(1,n+1):
        name = input("enter your name :")
        address= input("enter your address:")
        mobile= int(input("enter your mobile number:"))
        email= input("enter your email:")

        writer.writerow([name,address,mobile,email])


```

```python

# 4) Perform diffrent select operations
import sqlite3
conn=sqlite3.connect("college.db")
sql='select * from student'
data1=conn.execute(sql)
for row in data1:
    print(row)


sql2='select * from course'
data2=conn.execute(sql2)
for row in data2:
    print(row)


sql3='select * from student'
data3=conn.execute(sql3)
for row in data3:
    print(row)


sql4='select * from student where branch ="cse"'
data4=conn.execute(sql4)
for row in data4:
    print(row)

conn.close()
```
