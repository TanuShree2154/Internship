```python
# 1) Practise Pandas Series
# 	Create a Pandas Series from Dictionary 

import pandas as pd
data={1:'c++',2:'c',3:'python',4:'java',5:'sql'}
ps=pd.Series(data)

# 	Create a Pandas Series from Lists 
data1=['pandas','matplotlib','seaborn','numpy']
ps1=pd.Series(data1, index=['i1','i2','i3','i4'])

# 	Access the elements of a Series in Pandas  
print(ps)
print(ps1)
print(ps[2])
print(ps1['i1'])

```

```python
# 2) DataFrames
# 	Make a Pandas DataFrame with a two-dimensional Python list 
import pandas as pd
data={
    "Name":['tanu','labhee','isahni','paridhi'],
    "branch":['cse','cse','cse','ece']
}
ps=pd.DataFrame(data)
print(ps)
# 	Create DataFrame from Python dict 
data1={
    "Name":{1:'tanu',2:'labhee',3:'isahni',4:'paridhi'},
    "branch":{1:'cse',2:'cse',3:'cse',4:'ece'}
}
ps1=pd.DataFrame(data1)
print(ps1)
# 	Create Pandas dataframe using list of lists 
data2={
    'subjects':[['hindi','sanskrit'],['english','french'],['science','maths']],
    'marks':[[10,20],[30,40],[50,60]]
}
ps2=pd.DataFrame(data2)
print(ps2)
# 	Create a Pandas dataframe using list of tuples 
data3={
    'subjects':[('hindi','sanskrit'),('english','french'),('science','maths')],
    'marks':[(10,20),(30,40),(50,60)]
}
ps3=pd.DataFrame(data3)
print(ps3)
# 	Create a Pandas DataFrame from List of Dicts 
data4={
    'subjects':[{1:'hindi',2:'sanskrit'},{3:'english',4:'french'},{5:'science',6:'maths'}],
    'marks':[{1:10,2:20},{3:30,4:40},{5:50,6:60}]
}
ps4=pd.DataFrame(data4)
print(ps4)
```

```python
# 3)Data iteration
# 	Different ways to iterate over rows in Pandas Dataframe 
import pandas as pd
data={
    "Name":['tanu','labhee','isahni','paridhi'],
    "branch":['cse','cse','cse','ece']
}
ps=pd.DataFrame(data)
print(ps)
ps.index=[1,2,3,4]
ps.columns=['col1','col2']
print(ps.loc[1:3])
print(ps.loc[1:3,'col1'])
# 	Selecting rows in pandas DataFrame based on conditions 
print(ps.loc[1,'col1':'col2'])
# 	Select any row from a Dataframe using iloc[]
print(ps.iloc[1])
# 	Limited rows selection with given column
print(ps.loc[1:3])
# 	Drop rows from the dataframe based on certain condition applied on a column 
res=ps.drop([2])
print(res)
# 	Insert row at given position in Pandas Dataframe 
new_row = pd.DataFrame({"Name":["ritu"], "branch":["ece"]})
ps = pd.concat([ps.iloc[:1], new_row, ps.iloc[1:]]).reset_index(drop=True)

print(ps)
# 	Create a list from rows in Pandas dataframe 
rows_list = ps.values.tolist()
print(rows_list)

```
