```python
#1) Replace Nan with 0 and Interchange 3 rows and 3 columns of 2D array [[6, -8, 73, -110], 
# [np.nan, -8, 0, 94]] 

import numpy as np
arr = np.array([[6, -8, 73, -110],
                [np.nan, -8, 0, 94]])
print(arr)
arr = np.nan_to_num(arr, nan=0)
print(arr)
print(arr.T)
```

```python
# 2) Move axes of 3D array to new positions 
import numpy as np

arr = np.array([[6, -8, 73, -110],
                [np.nan, -8, 0, 94]])
print(arr.shape)
new_arr = np.moveaxis(arr, 0, 1)
print(new_arr)
print(new_arr.shape)
```

```python
# 3) Replace NaN values with average of columns
import numpy as np

arr = np.array([[6, -8, 73, -110],
                [np.nan, -8, 0, 94]])

mean = np.nanmean(arr, axis=0)

inds = np.where(np.isnan(arr))

arr[inds] = np.take(mean, inds[1])

print(arr)
```

```python
# 4) Replace negative value with zero in numpy array using replace

import numpy as np
arr = np.array([[6, -8, 73, -110],
                [np.nan, -8, 0, 94]])
arr[arr < 0] = 0
print(arr)
```
