```python
# 1) Study the open weather API show more data in your API calling program


import requests

def weather_data(city):
    url=f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid=c35042f42c1e7676b17dc7a795ce2670&units=metric"
    try:
        response=requests.get(url)
        response.raise_for_status()
        data= response.json()
        print(f"temprature: {data['main']['temp']}".center(70))
        print(f"feels_like: {data['main']['feels_like']}".center(70))
        print(f"temp_min: {data['main']['temp_min']}".center(70))
        print(f"temp_max: {data['main']['temp_max']}".center(70))
        print(f"pressure: {data['main']['pressure']}".center(70))
        print(f"humidity: {data['main']['humidity']}".center(70))
    except requests.execptions.RequestException as e:
        print(e)

city=input("enter city name:")
print(f"weather details of '{city}':")
weather_data(city)
```

```python
# 2) Try building today's game yourself

import random

l=['rock','paper','sicssor']
print("""
choose your choice:
      1. play
      0. exit
 """)
while True:
    p=int(input("enter want to play or not:"))
    if(p==1):
        uscore=0
        cscore=0
        for i in range(1,6):
            print("""
            choose your choice:
            1.rock 
            2.paper
            3.sicssor
            """)
            
            uip= int(input("enter your choice:"))
            if(uip==1):
                uch='rock'
            elif(uip==2):
                uch='paper'
            elif(uip==3):
                uch='sicssor'
            else:
                print("invalid input")
            cch=random.choice(l)

            print(f"your choice is {uch} and computer choice is {cch}")
            if(cch==uch):
                print("draw")
                cscore+=1
                uscore+=1
            elif((uch=="rock" and cch=="scissor") or
                (uch=="paper" and cch=="rock") or
                (uch=="scissor" and cch=="paper")):
                uscore+=2
                print("you win")
            else:
                cscore+=2
                print("computer win")
        print("Final resultss:")
        if uscore>cscore:
            print("you wonnn!!!!!")
        elif uscore<cscore:
            print("computer wonn!!")
        else:
            print("Game drawn")
            
    elif(p==0):
        print("game exited")
    else:
        print("invalid entry")     
```

```python
# 3) Search for more free API's generate your and call them to fetch data. Display some data in your 
# program.

import requests

def cat():
    url = "https://catfact.ninja/breeds"

    response = requests.get(url)
    response.raise_for_status()

    data = response.json()

    breeds = data["data"]

    # show first 5 breeds
    for i in range(5):
        b = breeds[i]
        print(f"Breed: {b['breed']}")
        print(f"Country: {b['country']}")
        print(f"Origin: {b['origin']}")
        

cat()
```
