```python
# > Create a 4×4×3 array and extract value at its second set, first row and last column 

arr = np.arange(48).reshape(4, 4, 3)
print(arr)
print(arr[1, 0, 2])
```

```python
# > Create a 4×4 and Extract Odd Rows and Even Columns 
arr=np.arange(16).reshape(4,4)
print(arr)

result = arr[1::2, ::2]

print("Odd Rows and Even Columns:")
print(result)
```

```python
# > Slice the first two rows and first two columns of econd set from a 4×4×3 array 
arr = np.arange(48).reshape(4, 4, 3)
result = arr[1, :2, :2]
print(result)
```

```python
# > Replace all odd numbers in a NumPy array with -1 by itterating using for loop 
# [[23, 56, 78, 93], [71, 82,13, 24]] 
arr = np.array([[23, 56, 78, 93],
                [71, 82, 13, 24]])

for i in range(arr.shape[0]):
    for j in range(arr.shape[1]):
        if arr[i][j] % 2 != 0:
            arr[i][j] = -1

print(arr)
```
