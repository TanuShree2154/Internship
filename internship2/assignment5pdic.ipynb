```python
#4) Write a Python program to calculate the total, highest value, and lowest value from a tuple
#  of integers without using the built-in sum(), max(), and min() functions. 

t = (10, 25, 5, 40, 15)

total = 0
highest = t[0]
lowest = t[0]

for num in t:
    total += num

    if num > highest:
        highest = num

    if num < lowest:
        lowest = num

print("Tuple:", t)
print("Total:", total)
print("Highest Value:", highest)
print("Lowest Value:", lowest)
```

```python
#5) Write a Python program to filter a tuple . n = (3, 14, 7, 22, 9, 41, 18, 5), keep only 
# values greater than 10 
n = (3, 14, 7, 22, 9, 41, 18, 5)
res= ()

for i in n:
    if i > 10:
        res+= (i,)

print("Original Tuple:", n)
print("Filtered Tuple:", res)
```
