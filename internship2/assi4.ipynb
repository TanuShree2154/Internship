```python
# > Create function for following
# 1. Write a Python function to find the maximum of three numbers.

def max_of_3(a,b,c):
    if(a>b and a>c):
        print(f"{a} is maximum")
    elif(b>c):
        print(f"{b} is maximum")
    else:
        print(f"{c} is maximum")

n1 = int(input("enter first number:"))
n2= int(input("enter second number:"))
n3= int(input("enter third number:"))
print(f"Maximum no. between {n1},{n2},{n3} is:")
max_of_3(n1,n2,n3)


```

```python
# 2. Write a Python function that takes a list and returns a new list with distinct elements from the
#  first list.

def new_list(l):
    result=[]
    for item in l:
        if item not in result:
            result.append(item)
    return result


l=[]
n=int(input("enter size of list:"))
for i in range(n):
    e= input(f"enter {i} elemnt in list")
    l.append(e)
print(new_list(l))
```

```python
# 3. Write a Python function to multiply all the numbers in a list.


def list_multiply(l):
    m=1
    for item in l:
        m=m*item
    return m


l=[]
n=int(input("enter size of list:"))
for i in range(n):
    e= int(input(f"enter {i} elemnt in list"))
    l.append(e)
print(list_multiply(l))
```

```python
# 4. Write a Python function to calculate the factorial of a number (a non-negative integer). 
# The function accepts the number as an argument.

def factorial(n):
    f=1
    for i in range(1,n+1):
        f=f*i
        i+=1
    return(f)


num= int(input("enter number to find factorial:"))
print(f"factorial of {num} is {factorial(num)}")
```

```python
# 5. Write a Python program to reverse a string.

def reverse_string(s):
    return s[-1::-1]

str=input("enter a string")
print(f"Reverse of string {str} is: {reverse_string(str)} ")
```

```python
# 6. Write a Python function to check whether a number falls within a given range.

def check_range(s,e,n):
    if(n>=s and n<=e):
        print(f"{n} lies between range({s},{e})")
    else:
        print(f"{n} not lies between range({s},{e})")


start= int(input("enter start of a range:"))
end=int(input("enter end id a range:"))
num = int(input("enter number to check range:"))
check_range(start, end , num)
```

```python
# 7. Write a Python function to Print Even Numbers from a Given List


def even_list(l):
    for item in l:
        if item%2 == 0:
            print(item)




l=[]
n=int(input("enter size of list:"))
for i in range(n):
    e= int(input(f"enter {i} element in list"))
    l.append(e)
even_list(l)
```

```python
# 8. Write a Python function that takes a number as a parameter and checks whether the number is prime
#  or not.

def check_prime(n):
    if(n<=1):
        print(f"{n} is not prime")
        return
    
    for i in range(2,n):
        if n%i == 0:
            print(f"{n} is not prime")
            return

    print(f"{n} is prime")


num=int(input("enter a number:"))
check_prime(num)
```

```python
# 9. Write a Python function that accepts a string and counts the number of upper and lower case letters. 

def count(s):
    u=0
    l=0
    for i in s:
        if(i>='a' and i<='z'):
            l+=1
        elif(i>='A' and i<='Z'):
            u+=1
    print(f"upper case letters are {u} and lower case letters are {l} in {s}")


str=input("enter a string:")
count(str)
```
