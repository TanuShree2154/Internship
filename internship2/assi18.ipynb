```python
# Create a PySpark application that generates a DataFrame containing 5 million records using spark.range(). Display the initial number of partitions,
# then increase the partitions to 12 using repartition(). After that, reduce the partitions to 3 using coalesce()
```

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("partition&coalesce").getOrCreate()
```

```python
df= spark.range(0,5000000)
```

```python
beforeRepartitioning = df.rdd.getNumPartitions()
```

```python
print("initial partitions: ", beforeRepartitioning )
```

```python
df_repartition = df.repartition(12)
```

```python
print("after Repartition: ", df_repartition.rdd.getNumPartitions())
```

```python
coalesce_partition = df_repartition.coalesce(3)
```

```python
print("After coalesce: ", coalesce_partition.rdd.getNumPartitions())
```

```python
spark.stop()
```
