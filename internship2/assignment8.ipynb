```python
# 2)Create two DataFrames, df1 and df2, with a common column (e.g., 'ID'). 
import pandas as pd
data1={
    'ID':[1,2,3,4,7],
    'emp':['emp1','emp2','emp3','emp4','emp7'],
    'age':[23,25,26,24,45]
}
df1=pd.DataFrame(data1)
# print(df1)
data2={
    'ID':[3,4,5,6],
    'emp':['emp3','emp4','emp5','emp6'],
    'age':[23,25,26,24]
}
df2=pd.DataFrame(data2)
# print(df2)
# >Perform an inner merge on this common column and display the resulting DataFrame.
res=df1.merge(df2 ,on='ID',how='inner')
print(res)
# >Perform a left join of df1 and df2 on the 'ID' column. Explain how missing values are handled 
# in the resulting DataFrame. Right Join and Index-Based Join.
df1_i = df1.set_index('ID')
df2_i = df2.set_index('ID')
res1=df1_i.join(df2_i,on='ID',how='left',lsuffix='left',rsuffix='right')
print(res1)
print("right join:\n")
res2=df1_i.join(df2_i,on='ID',how='right',lsuffix='left',rsuffix='right')
print(res2)
print("index_based join:\n")
res3=df1_i.join(df2_i,on='ID',lsuffix='left',rsuffix='right')
print(res3)


# >Perform a right join using pd.merge() on a common column, then perform a join using df.join()
#  based on the index. Compare the results. Merging with Multiple Keys.
right_merge = pd.merge(df1, df2, on='ID', how='right')
print("right merge:\n")
print(right_merge)

join_result = df1_i.join(df2_i, how='right', lsuffix='_df1', rsuffix='_df2')
print("right join:\n")
print(join_result)
#comparision 
# merge uses colums(ID) and join uses index only
```

```python
# 3) Create three DataFrames. Vertically concatenate two of them using pd.concat(), 
# then merge the resulting DataFrame with the third DataFrame on a common key.
#  T Understand join() vs. merge().
import pandas as pd
a=pd.DataFrame({
    "Name":['a1','a2','a3','a4'],
    "subject":['sub1','sub2','sub3','sub4'],
    "marks":[90,70,80,70]
},index=[1,2,3,4])
b=pd.DataFrame({
    "Name":['a5','a6','a7','a8'],
    "subject":['sub5','sub2','sub3','sub6'],
    "marks":[95,64,48,64]
},index=[1,2,3,4])
c=pd.DataFrame({
    "Name":['a5','a9','a7','a11'],
    "subject":['sub5','sub9','sub3','sub11'],
    "marks":[95,54,48,25]
},index=[1,2,3,4])

#concatination
res=pd.concat([a,b],ignore_index=True)
print("concatination:\n",res)
 #concat column-wise
res1=res.merge(c,on='Name',how='outer')
print("merge:\n",res1)


# Also Explain the primary differences between df.join() and pd.merge(),
#merge() joins DataFrames using columns like SQL joins, while join() combines DataFrames using 
# their index.


```
